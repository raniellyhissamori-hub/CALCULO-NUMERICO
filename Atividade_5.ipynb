```python
import numpy as np

def eliminacao_gauss(A, b):
    # Converte matrizes e vetores para tipo float para suportar divisões reais
    A = np.array(A, dtype=float)
    b = np.array(b, dtype=float)
    n = len(b)

    # Construção da matriz aumentada Ab = [A | b]
    Ab = np.hstack([A, b.reshape(-1, 1)])

    # 1. Eliminação Gaussiana (triangularização)
    for k in range(n - 1):
        for i in range(k + 1, n):
            fator = Ab[i, k] / Ab[k, k]
            Ab[i, k:] -= fator * Ab[k, k:]

    # 2. Substituição Regressiva
    x = np.zeros(n)
    for i in range(n - 1, -1, -1):
        soma = np.sum(Ab[i, i + 1:n] * x[i + 1:n])
        x[i] = (Ab[i, n] - soma) / Ab[i, i]

    return x

# --- Teste com o exercício 2 ---
A_ex2 = np.array([[2, 1, -1],
                  [-3, -1, 2],
                  [-2, 1, 2]], dtype=float)

b_ex2 = np.array([8, -11, -3], dtype=float)

# Execução do algoritmo customizado
sol_custom = eliminacao_gauss(A_ex2, b_ex2)

# Comparação com o resolvedor do NumPy
sol_numpy = np.linalg.solve(A_ex2, b_ex2)

print("Solução via Eliminação de Gauss (custom):", sol_custom)
print("Solução via numpy.linalg.solve:", sol_numpy)
print("Diferença máxima:", np.max(np.abs(sol_custom - sol_numpy)))
```
